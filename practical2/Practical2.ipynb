```python
import pandas as pd
import sqlite3
import json
import random
import logging
from faker import Faker
from datetime import datetime
```

```python
fake = Faker()
```

```python
print("Name:", fake.name())
print("Email:", fake.email())
print("City:", fake.city())
print("Phone:", fake.phone_number())
```

```python
customers = []
```

```python
for i in range(100):
    customers.append({
        "customer_id": i + 1,
        "name": fake.name(),
        "email": fake.email(),
        "phone": fake.phone_number(),
        "city": fake.city(),
        "state": fake.state(),
        "country": fake.country(),
        "age": random.randint(18, 60),
        "registration_date": fake.date_between(start_date='-2y', end_date='today')
    })
```

```python
df_customers = pd.DataFrame(customers)
```

```python
df_customers.to_csv("customer_data.csv", index=False)
```

```python
df_dirty = df_customers.copy()
```

```python
# Missing primary key
df_dirty.loc[5, "customer_id"] = None

# Missing email
df_dirty.loc[10, "email"] = None

# Invalid email format
df_dirty.loc[15, "email"] = "john.gmail.com"

# Invalid phone number
df_dirty.loc[20, "phone"] = "ABCDE12345"

# Duplicate record
df_dirty = pd.concat([df_dirty, df_dirty.iloc[[25]]], ignore_index=True)

df_dirty.head(30)
```

```python
df_dirty.isnull().sum()
```

```python
df_dirty.duplicated().sum()
```

```python
df_dirty.loc[[5, 10, 15, 20, 100]]
```

```python
missing_id = df_dirty[df_dirty["customer_id"].isnull()]

print("Missing Primary Keys:")
missing_id
```

```python
missing_email = df_dirty[df_dirty["email"].isnull()]

print("Missing Emails:")
missing_email
```

```python
invalid_email = df_dirty[
    ~df_dirty["email"].fillna("").str.contains("@")
]

print("Invalid Email Addresses:")
invalid_email
```

```python
invalid_phone = df_dirty[
    df_dirty["phone"].str.match(r"^[A-Za-z]", na=False)
]

print("Invalid Phone Numbers:")
invalid_phone
```

```python
duplicates = df_dirty[df_dirty.duplicated()]

print("Duplicate Records:")
duplicates
```

```python
print("========== DATA QUALITY REPORT ==========")

print(f"Total Records          : {len(df_dirty)}")
print(f"Missing Customer IDs   : {missing_id.shape[0]}")
print(f"Missing Emails         : {missing_email.shape[0]}")
print(f"Invalid Emails         : {invalid_email.shape[0]}")
print(f"Invalid Phone Numbers  : {invalid_phone.shape[0]}")
print(f"Duplicate Records      : {duplicates.shape[0]}")
```

```python
transactions = []

for i in range(50):
    transaction = {
        "transaction_id": i + 1,
        "customer": {
            "customer_id": random.randint(1, 100),
            "name": fake.name()
        },
        "payment": {
            "amount": round(random.uniform(100, 5000), 2),
            "method": random.choice(["UPI", "Credit Card", "Debit Card", "Net Banking"])
        },
        "timestamp": str(datetime.now())
    }

    transactions.append(transaction)

transactions[:2]
```

```python
with open("transactions.json", "w") as file:
    json.dump(transactions, file, indent=4)

print("transactions.json created successfully.")
```

```python
with open("transactions.json", "r") as file:
    print(file.read()[:500])
```

```python
transactions[0]["payment_details"] = transactions[0].pop("payment")
```

```python
if "payment" not in transactions[0]:
    print("⚠️ Schema Drift Detected!")
    print("'payment' field has changed to 'payment_details'.")
```

```python
config = """
HOST=localhost
PORT=5432
DATABASE=customer_db
USERNAME=admin
PASSWORD=
TIMEOUT=abc
"""

with open("config.txt", "w") as file:
    file.write(config)

print("config.txt created successfully.")
```

```python
with open("config.txt", "r") as file:
    config_lines = file.readlines()

print("Configuration Validation:")

for line in config_lines:
    if "=" in line:
        key, value = line.strip().split("=", 1)

        if value == "":
            print(f"Missing value for {key}")

        if key == "TIMEOUT" and not value.isdigit():
            print("TIMEOUT should be numeric")
```

```python
conn = sqlite3.connect("customer.db")

df_dirty.to_sql(
    "customers",
    conn,
    if_exists="replace",
    index=False
)

print("Customer data stored successfully!")
```

```python
query = pd.read_sql("SELECT * FROM customers LIMIT 5", conn)

query
```
